## calculate edge closeness for all edges of the filtered subjects

According to Bröhl & Lehnertz (2022), edge closeness centrality $C_e^C$ for an edge $k = (a, b)$ is defined as:
$$C_e^C(k) = (E - 1) \frac{C_v^C(a) C_v^C(b)}{C_v^C(a) + C_v^C(b)}$$
where $C_v^C(a)$ and $C_v^C(b)$ are the node closeness centralities of endpoints $a$ and $b$ computed using shortest path distances ($d = 1 / |r|$), and $E$ is the total number of edges in the network.


```python
import networkx as nx
import numpy as np


def compute_edge_closeness_centrality(G, distance="distance"):
    node_cc = nx.closeness_centrality(G, distance=distance)
    E = G.number_of_edges()
    ec = {}
    for u, v in G.edges():
        c_u = node_cc.get(u, 0.0)
        c_v = node_cc.get(v, 0.0)
        if c_u + c_v > 0:
            ec[(u, v)] = (E - 1) * (c_u * c_v) / (c_u + c_v)
        else:
            ec[(u, v)] = 0.0
    return ec


subject_edge_closeness = {}

for subj, df in subject_matrices.items():
    # Create graph from adjacency matrix
    G = nx.from_pandas_adjacency(df)
    
    # Calculate distance for shortest path computation: distance = 1 / correlation
    for u, v, d in G.edges(data=True):
        weight = d.get("weight", 0)
        if weight > 0:
            d["distance"] = 1.0 / weight
        else:
            d["distance"] = np.inf

    # Calculate edge closeness centrality
    ec = compute_edge_closeness_centrality(G, distance="distance")
    
    # Store the result
    subject_edge_closeness[subj] = ec

print(f"Calculated edge closeness for {len(subject_edge_closeness)} subjects")
```

## calculate weighted jaccard index for edge closeness

This cell calculates the weighted Jaccard index (min/max) per segment for all 171 edges between every pair of subjects. 
For each edge, the Jaccard index is computed for each of the 12 segments by dividing the minimum closeness value by the maximum value. These 12 values are then averaged to give exactly 171 Jaccard indices per subject pair (one for each edge). Finally, the average of these 171 indices is also computed.


```python
import networkx as nx
import numpy as np
import pandas as pd
from itertools import combinations
from pathlib import Path


def compute_edge_closeness_centrality(G, distance='distance'):
    node_cc = nx.closeness_centrality(G, distance=distance)
    E = G.number_of_edges()
    ec = {}
    for u, v in G.edges():
        c_u = node_cc.get(u, 0.0)
        c_v = node_cc.get(v, 0.0)
        if c_u + c_v > 0:
            ec[(u, v)] = (E - 1) * (c_u * c_v) / (c_u + c_v)
        else:
            ec[(u, v)] = 0.0
    return ec


# 1. Calculate edge closeness for each segment individually for all filtered subjects
subject_segment_ec = {}

for subj in formatted_subjects:
    subj_dir = abs_segment_base / subj / "ses-1"
    if not subj_dir.exists():
        continue
        
    csv_files = sorted(subj_dir.glob('*.csv'))
    
    subject_segment_ec[subj] = []
    
    for csv_file in csv_files[:n_segments]:
        df = pd.read_csv(csv_file, index_col=0)
        G = nx.from_pandas_adjacency(df)
        
        # Calculate distance: 1 / correlation
        for u, v, d in G.edges(data=True):
            weight = d.get('weight', 0)
            if weight > 0:
                d['distance'] = 1.0 / weight
            else:
                d['distance'] = np.inf
                
        # Calculate edge closeness for this segment
        ec = compute_edge_closeness_centrality(G, distance='distance')
        
        # Standardize edge tuple order to ensure consistency
        normalized_ec = {}
        for (u, v), val in ec.items():
            edge_tuple = tuple(sorted([u, v]))
            normalized_ec[edge_tuple] = val
            
        subject_segment_ec[subj].append(normalized_ec)

print(f"Calculated individual segment edge closeness for {len(subject_segment_ec)} subjects.")

# 2. Calculate Weighted Jaccard Index for all pairs of subjects
sample_subj = formatted_subjects[0]
edges_list = list(subject_segment_ec[sample_subj][0].keys())

pair_jaccard_avg = {}
pair_jaccard_all_edges = {}

# Iterate over all 20C2 pairs
for subj_A, subj_B in combinations(formatted_subjects, 2):
    edge_jaccards = []
    
    for e in edges_list:
        segment_jaccards = []
        
        for s in range(n_segments):
            if s < len(subject_segment_ec[subj_A]) and s < len(subject_segment_ec[subj_B]):
                val_A = subject_segment_ec[subj_A][s].get(e, 0)
                val_B = subject_segment_ec[subj_B][s].get(e, 0)
                
                if max(val_A, val_B) > 0:
                    jaccard_s = min(val_A, val_B) / max(val_A, val_B)
                else:
                    jaccard_s = 0
                segment_jaccards.append(jaccard_s)
        
        # Average the segment Jaccards for this edge
        edge_jaccard_avg = np.mean(segment_jaccards) if segment_jaccards else 0
        edge_jaccards.append(edge_jaccard_avg)
        
    pair_jaccard_all_edges[(subj_A, subj_B)] = edge_jaccards
    pair_jaccard_avg[(subj_A, subj_B)] = np.mean(edge_jaccards)

print(f"Calculated average Jaccard index for {len(pair_jaccard_avg)} subject pairs.")
```

## Plot Jaccard Index Heatmap for Subject Pairs

Plots the Jaccard index for all 171 edges across all 190 subject pairs. The y-axis displays subject pairs sorted by the difference in their ages, with the smallest age difference at the bottom.


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Get the age data
ages_df = pd.read_csv('../age_acq_time_sorted.csv')
target_labels = [f"{subj}_ses-1" for subj in formatted_subjects]
ages_filtered = ages_df[ages_df['label'].isin(target_labels)]

age_dict = {}
for idx, row in ages_filtered.iterrows():
    subj = row['label'].split('_')[0]
    age_dict[subj] = row['age_acq_time']

# Calculate age difference for each pair
pair_age_diff = {}
for pair in pair_jaccard_all_edges.keys():
    subj_A, subj_B = pair
    diff = abs(age_dict[subj_A] - age_dict[subj_B])
    pair_age_diff[pair] = diff

# Sort pairs by age difference descending (largest diff at top, smallest diff at bottom)
sorted_pairs = sorted(pair_age_diff.keys(), key=lambda x: pair_age_diff[x], reverse=True)

# Map original edges_list to edge_labels order
edge_to_index = {e: i for i, e in enumerate(edges_list)}

heatmap_data = []
yticklabels = []

for pair in sorted_pairs:
    p0 = str(int(pair[0].replace('sub-NORB', '')))
    p1 = str(int(pair[1].replace('sub-NORB', '')))
    yticklabels.append(f"{p0}-{p1}")
    row_data = []
    for e in edge_labels:
        u, v = e.split('-')
        edge_tuple = tuple(sorted([u, v]))
        idx = edge_to_index[edge_tuple]
        val = pair_jaccard_all_edges[pair][idx]
        row_data.append(val)
    heatmap_data.append(row_data)

heatmap_data = np.array(heatmap_data)

# Plot heatmap
plt.figure(figsize=(50, 20))
sns.heatmap(heatmap_data, cmap='hot', xticklabels=edge_labels, yticklabels=yticklabels)

# labels of the region pair at the top of the heatmap
import matplotlib.transforms as transforms

ax = plt.gca()
# Combine X in data coordinates (columns) with Y in axes coordinates (plot box height)
trans = transforms.blended_transform_factory(ax.transData, ax.transAxes)

start_idx = 0
for i in range(1, len(edge_labels) + 1):
    r_prev = f"{edge_labels[i-1].split('-')[0][0]}-{edge_labels[i-1].split('-')[1][0]}"
    r_curr = f"{edge_labels[i].split('-')[0][0]}-{edge_labels[i].split('-')[1][0]}" if i < len(edge_labels) else None
    
    if r_prev != r_curr:
        # 1. Draw vertical boundary line between region groups
        if i < len(edge_labels):
            ax.axvline(x=i, color='cyan', linestyle='--', linewidth=1.5)
            
        # 2. Calculate horizontal center of this region group
        mid_x = (start_idx + i) / 2.0
        
        # 3. Place label slightly above the top edge of the plot (y=1.01 in axes coordinates)
        ax.text(mid_x, 1.01, r_prev, transform=trans, ha='center', va='bottom', fontsize=10, fontweight='bold')
        
        start_idx = i


plt.title('Subject Pair Edge Closeness Jaccard Index', pad=35)
plt.xlabel('Edges')
plt.ylabel('Subject Pairs (Largest to Smallest Age Difference)')
plt.xticks(rotation=90, fontsize=8)
plt.yticks(fontsize=8)
plt.tight_layout()
plt.show()
```

## Region-wise Statistical Summary of Edge Closeness Jaccard Index

This cell calculates and displays a grouped bar plot showing the percentage of Edge Closeness Jaccard index values exceeding **0.75** and **0.90** for each specific brain region connection type (F-F, F-C, F-T, ..., O-O).


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# --- 1. Load Hemisphere Labels ---
labels_df = pd.read_csv('../electrode_labels.csv') 
hemi_map = dict(zip(labels_df['channel'], labels_df['label']))

def get_hemi_pair(edge):
    u, v = edge.split('-')
    # Default to 'C' if missing just in case
    h1 = hemi_map.get(u, 'C')
    h2 = hemi_map.get(v, 'C')
    
    # Standardize order so L-R and R-L both become L-R
    order = {'L': 0, 'C': 1, 'R': 2}
    if order.get(h1, 99) <= order.get(h2, 99):
        pair = f"{h1}-{h2}"
    else:
        pair = f"{h2}-{h1}"
    
    # Rename C-R to R-C per your request
    if pair == 'C-R':
        pair = 'R-C'
    return pair

# --- 2. Define region order and canonical mapping ---
region_order = {'F': 0, 'C': 1, 'T': 2, 'P': 3, 'O': 4}

def get_region_pair(edge):
    u, v = edge.split('-')
    r1, r2 = u[0], v[0]
    if region_order.get(r1, 99) <= region_order.get(r2, 99):
        return f"{r1}-{r2}"
    else:
        return f"{r2}-{r1}"

region_groups = ['F-F', 'F-C', 'F-T', 'F-P', 'F-O',
                 'C-C', 'C-T', 'C-P', 'C-O',
                 'T-T', 'T-P', 'T-O',
                 'P-P', 'P-O',
                 'O-O']

hemi_groups = ['L-L', 'L-C', 'L-R', 'R-R', 'R-C', 'C-C']

# --- 3. Calculate percentages grouped by Region AND Hemisphere ---
stats_data = []

for cat in region_groups:
    # Find all edges belonging to this Region Pair
    edge_indices = [idx for idx, e in enumerate(edge_labels) if get_region_pair(e) == cat]
    
    # Total values across all subject pairs for this region (denominator for percentage)
    total_vals_in_region = len(edge_indices) * heatmap_data.shape[0]
    
    for hemi in hemi_groups:
        # Filter those edges further by Hemisphere Pair
        hemi_indices = [idx for idx in edge_indices if get_hemi_pair(edge_labels[idx]) == hemi]
        
        if hemi_indices and total_vals_in_region > 0:
            vals = np.asarray(heatmap_data)[:, hemi_indices].flatten()
            # Calculate proportion relative to the entire Region Pair
            pct_30 = (np.sum(vals > 0.30) / total_vals_in_region) * 100.0
            pct_50 = (np.sum(vals > 0.50) / total_vals_in_region) * 100.0
            pct_75 = (np.sum(vals > 0.75) / total_vals_in_region) * 100.0
            pct_90 = (np.sum(vals > 0.90) / total_vals_in_region) * 100.0
        else:
            pct_30, pct_50, pct_75, pct_90 = 0.0, 0.0, 0.0, 0.0
            
        stats_data.append({'Region Pair': cat, 'Threshold': '> 0.30', 'Hemi': hemi, 'Percentage': pct_30})
        stats_data.append({'Region Pair': cat, 'Threshold': '> 0.50', 'Hemi': hemi, 'Percentage': pct_50})
        stats_data.append({'Region Pair': cat, 'Threshold': '> 0.75', 'Hemi': hemi, 'Percentage': pct_75})
        stats_data.append({'Region Pair': cat, 'Threshold': '> 0.90', 'Hemi': hemi, 'Percentage': pct_90})

df_stats = pd.DataFrame(stats_data)

# --- 4. Plot the Stacked Bar Charts ---
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 14))

# Pick 6 distinct colors for the 6 hemisphere regions
colors = sns.color_palette("Set2", 6) 
color_map = dict(zip(hemi_groups, colors))

x = np.arange(len(region_groups))
width = 0.35

def plot_grouped_stacked(ax, thresh1, thresh2, title):
    # thresh1
    df1 = df_stats[df_stats['Threshold'] == thresh1].pivot(index='Region Pair', columns='Hemi', values='Percentage')
    df1 = df1.reindex(index=region_groups, columns=hemi_groups).fillna(0)
    
    # thresh2
    df2 = df_stats[df_stats['Threshold'] == thresh2].pivot(index='Region Pair', columns='Hemi', values='Percentage')
    df2 = df2.reindex(index=region_groups, columns=hemi_groups).fillna(0)
    
    bottom1 = np.zeros(len(region_groups))
    bottom2 = np.zeros(len(region_groups))
    
    for hemi in hemi_groups:
        vals1 = df1[hemi].values
        vals2 = df2[hemi].values
        
        # Plot thresh1 on left
        bars1 = ax.bar(x - width/2, vals1, width, bottom=bottom1, label=hemi, color=color_map[hemi], edgecolor='white', linewidth=0.5)
        # Plot thresh2 on right (no label to avoid legend duplicates)
        bars2 = ax.bar(x + width/2, vals2, width, bottom=bottom2, color=color_map[hemi], edgecolor='white', linewidth=0.5)
        
        # Add labels
        ax.bar_label(bars1, labels=[f'{v:.1f}%' if v > 2 else '' for v in vals1], label_type='center', fontsize=8)
        ax.bar_label(bars2, labels=[f'{v:.1f}%' if v > 2 else '' for v in vals2], label_type='center', fontsize=8)
        
        bottom1 += vals1
        bottom2 += vals2

    ax.set_xticks(x)
    ax.set_xticklabels(region_groups)
    ax.set_title(title, fontsize=14)
    ax.set_ylabel('Percentage of Values (%)', fontsize=12)
    ax.set_ylim(0, 100)
    ax.legend(title='Hemisphere Connection', loc='upper right')

# --- Plot 1: > 0.75 and > 0.90 ---
plot_grouped_stacked(ax1, '> 0.75', '> 0.90', 'Percentage of Jaccard Index Values > 0.75 (Left) and > 0.90 (Right)')

# --- Plot 2: > 0.50 and > 0.30 ---
plot_grouped_stacked(ax2, '> 0.50', '> 0.30', 'Percentage of Jaccard Index Values > 0.50 (Left) and > 0.30 (Right)')

ax2.set_xlabel('Brain Region Connections', fontsize=12)

plt.tight_layout()
plt.show()
```
