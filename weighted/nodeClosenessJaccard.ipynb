## import the segment wise absolute correlation matrices

```python
from pathlib import Path
import pandas as pd

abs_segment_base = Path('../correlation_matrices_segment_wise_absolute')
matrix_files = list(abs_segment_base.rglob('*.csv'))

if matrix_files:
    sample = pd.read_csv(matrix_files[0], index_col=0)
    print(f"Found {len(matrix_files)} absolute matrices. Sample shape: {sample.shape}")
```

## filter the subjects for our study

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from itertools import combinations

# Filter subjects
subject_ids = [64, 65, 69, 70, 75, 77, 78, 79, 80, 81, 82, 83, 84, 86, 87, 89, 90, 94, 96, 103]
formatted_subjects = [f"sub-NORB{sid:05d}" for sid in subject_ids]

# Extract first 12 segments for session 1
n_segments = 12
subject_matrices = {}

for subj in formatted_subjects:
    subj_dir = abs_segment_base / subj / "ses-1"
    if not subj_dir.exists():
        print(f"Warning: directory not found for {subj}")
        continue
        
    csv_files = sorted(subj_dir.glob('*.csv'))
    if len(csv_files) < n_segments:
        print(f"Warning: {subj} has less than {n_segments} segments")
        
    matrices = []
    last_df = None
    for csv_file in csv_files[:n_segments]:
        df = pd.read_csv(csv_file, index_col=0)
        last_df = df
        matrices.append(df.values)
        
    if matrices:
        avg_matrix = np.mean(matrices, axis=0)
        if 'last_df' in locals() and last_df is not None:
            subject_matrices[subj] = pd.DataFrame(avg_matrix, index=last_df.index, columns=last_df.columns)
        
print(f"Processed {len(subject_matrices)} subjects")
```

## calculate node closeness for all nodes of the filtered subjects

```python
import networkx as nx
import numpy as np

subject_node_closeness = {}

for subj, df in subject_matrices.items():
    # Create graph from adjacency matrix
    G = nx.from_pandas_adjacency(df)
    
    # Calculate distance for shortest path computation: distance = 1 / correlation
    for u, v, d in G.edges(data=True):
        weight = d.get('weight', 0)
        if weight > 0:
            d['distance'] = 1.0 / weight
        else:
            d['distance'] = np.inf

    # Calculate node closeness centrality using 'distance' as edge distance
    nc = nx.closeness_centrality(G, distance='distance')
    
    # Store the result
    subject_node_closeness[subj] = nc

print(f"Calculated node closeness for {len(subject_node_closeness)} subjects")
```

## plot node closeness heatmap

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

node_labels = ['Fp1', 'Fp2', 'F3', 'F4', 'F7', 'F8', 'FZ', 'C3', 'C4', 'CZ', 'T3', 'T4', 'T5', 'T6', 'P3', 'P4', 'PZ', 'O1', 'O2']

# Load ages and sort subjects increasing (oldest top, youngest bottom)
ages_df = pd.read_csv('../age_acq_time_sorted.csv')
target_labels = [f"{subj}_ses-1" for subj in formatted_subjects]
ages_filtered = ages_df[ages_df['label'].isin(target_labels)].copy()
# User requested increasing order of their age
assert isinstance(ages_filtered, pd.DataFrame)
ages_filtered = ages_filtered.sort_values(by='age_acq_time', ascending=False)

sorted_subject_labels = ages_filtered['label'].tolist()
sorted_subjects = [label.split('_')[0] for label in sorted_subject_labels]

# Prepare data for heatmap
heatmap_data = []
valid_subjects = []

for subj in sorted_subjects:
    if subj in subject_node_closeness:
        valid_subjects.append(subj)
        nc = subject_node_closeness[subj]
        row_data = []
        for n in node_labels:
            val = nc.get(n, 0)
            row_data.append(val)
        heatmap_data.append(row_data)

heatmap_data = np.array(heatmap_data)

# Plot heatmap
plt.figure(figsize=(20, 10))
sns.heatmap(heatmap_data, cmap='hot', xticklabels=node_labels, yticklabels=valid_subjects)
plt.title('Node Closeness Centrality')
plt.xlabel('Nodes')
plt.ylabel('Subjects (Youngest to Oldest)')
plt.show()
```

## calculate weighted jaccard index for node closeness

this cell calculates the weighted jaccard index (min/max) per segment for all 19 nodes between every pair of subjects.
for each node, the jaccard index is computed for each of the 12 segments by dividing the minimum closeness value by the maximum value. These 12 values are then averaged to give exactly 19 Jaccard indices per subject pair (one for each node). finally, the average of these 19 indices is also computed.

```python
import networkx as nx
import numpy as np
import pandas as pd
from itertools import combinations
from pathlib import Path

# 1. Calculate node closeness for each segment individually for all filtered subjects
subject_segment_nc = {}

for subj in formatted_subjects:
    subj_dir = abs_segment_base / subj / "ses-1"
    if not subj_dir.exists():
        continue

    csv_files = sorted(subj_dir.glob('*.csv'))
    subject_segment_nc[subj] = []

    for csv_file in csv_files[:n_segments]:
        df = pd.read_csv(csv_file, index_col=0)
        G = nx.from_pandas_adjacency(df)

        # Calculate distance: 1 / correlation
        for u, v, d in G.edges(data=True):
            weight = d.get('weight', 0)
            if weight > 0:
                d['distance'] = 1.0 / weight
            else:
                d['distance'] = np.inf

        # Calculate closeness for this segment
        nc = nx.closeness_centrality(G, distance='distance')
        subject_segment_nc[subj].append(nc)

print(f"Calculated individual segment node closeness for {len(subject_segment_nc)} subjects.")

# 2. Calculate Weighted Jaccard Index for all pairs of subjects
sample_subj = formatted_subjects[0]
nodes_list = list(subject_segment_nc[sample_subj][0].keys())

pair_jaccard_avg = {}
pair_jaccard_all_nodes = {}

# Iterate over all 20C2 pairs
for subj_A, subj_B in combinations(formatted_subjects, 2):
    node_jaccards = []

    for n in nodes_list:
        segment_jaccards = []

        for s in range(n_segments):
            if s < len(subject_segment_nc[subj_A]) and s < len(subject_segment_nc[subj_B]):
                val_A = subject_segment_nc[subj_A][s].get(n, 0)
                val_B = subject_segment_nc[subj_B][s].get(n, 0)

                if max(val_A, val_B) > 0:
                    jaccard_s = min(val_A, val_B) / max(val_A, val_B)
                else:
                    jaccard_s = 0
                segment_jaccards.append(jaccard_s)

        # Average the segment Jaccards for this node
        node_jaccard_avg = np.mean(segment_jaccards) if segment_jaccards else 0
        node_jaccards.append(node_jaccard_avg)

    pair_jaccard_all_nodes[(subj_A, subj_B)] = node_jaccards
    pair_jaccard_avg[(subj_A, subj_B)] = np.mean(node_jaccards)

print(f"Calculated average Jaccard index for {len(pair_jaccard_avg)} subject pairs.")
```

## plot jaccard index heatmap for subject pairs

plots the jaccard index for all 19 nodes across all 190 subject pairs. the y-axis displays subject pairs sorted by the difference in their ages, with the smallest age difference at the bottom.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Get the age data
ages_df = pd.read_csv('../age_acq_time_sorted.csv')
target_labels = [f"{subj}_ses-1" for subj in formatted_subjects]
ages_filtered = ages_df[ages_df['label'].isin(target_labels)]

age_dict = {}
for idx, row in ages_filtered.iterrows():
    subj = row['label'].split('_')[0]
    age_dict[subj] = row['age_acq_time']

# Calculate age difference for each pair
pair_age_diff = {}
for pair in pair_jaccard_all_nodes.keys():
    subj_A, subj_B = pair
    diff = abs(age_dict[subj_A] - age_dict[subj_B])
    pair_age_diff[pair] = diff

# Sort pairs by age difference descending (largest diff at top, smallest diff at bottom)
sorted_pairs = sorted(pair_age_diff.keys(), key=lambda x: pair_age_diff[x], reverse=True)

# Map original nodes_list to node_labels order
node_to_index = {n: i for i, n in enumerate(nodes_list)}

heatmap_data = []
yticklabels = []

for pair in sorted_pairs:
    p0 = str(int(pair[0].replace('sub-NORB', '')))
    p1 = str(int(pair[1].replace('sub-NORB', '')))
    yticklabels.append(f"{p0}-{p1}")
    row_data = []
    for n in node_labels:
        idx = node_to_index.get(n, -1)
        val = pair_jaccard_all_nodes[pair][idx] if idx != -1 else 0
        row_data.append(val)
    heatmap_data.append(row_data)

heatmap_data = np.array(heatmap_data)

# Plot heatmap
plt.figure(figsize=(50, 50))
sns.heatmap(heatmap_data, cmap='hot', xticklabels=node_labels, yticklabels=yticklabels)

# add label above each lobe
import matplotlib.transforms as transforms

ax = plt.gca()
trans = transforms.blended_transform_factory(ax.transData, ax.transAxes)

# Define each lobe label and its ending column index based on node_labels
# F: 7 nodes (0-6), C: 3 nodes (7-9), T: 4 nodes (10-13), P: 3 nodes (14-16), O: 2 nodes (17-18)
lobes = [
    ('F', 7),
    ('C', 10),
    ('T', 14),
    ('P', 17),
    ('O', 19)
]

start_idx = 0
for label, end_idx in lobes:
    # Draw vertical cyan dashed boundary line between lobes (except at the very right edge)
    if end_idx < 19:
        ax.axvline(x=end_idx, color='cyan', linewidth=2)
        
    # Add lobe label centered above its respective section at the top of the plot
    mid_x = (start_idx + end_idx) / 2.0
    ax.text(mid_x, 1.005, label, transform=trans, ha='center', va='bottom', fontsize=20, fontweight='bold')
    
    start_idx = end_idx


plt.title('Subject Pair Node Closeness Jaccard Index')
plt.xlabel('Nodes')
plt.ylabel('Subject Pairs (Largest to Smallest Age Difference)')
plt.xticks(rotation=90, fontsize=8)
plt.yticks(fontsize=8)
plt.tight_layout()
plt.show()
```

## Region-wise Statistical Summary of Node Closeness Jaccard Index

This cell calculates and displays a grouped bar plot showing the percentage of Node Closeness Jaccard index values exceeding **0.75** and **0.90** for each distinct brain region (**F**, **C**, **T**, **P**, and **O**).

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# --- 1. Load Hemisphere Labels ---
labels_df = pd.read_csv('../electrode_labels.csv')
hemi_map = dict(zip(labels_df['channel'], labels_df['label']))

def get_hemi(node):
    return hemi_map.get(node, 'C')

# --- 2. Map every node to Region and Hemisphere ---
region_groups = ['F', 'C', 'T', 'P', 'O']
hemi_groups = ['L', 'R', 'C']

stats_data = []

for cat in region_groups:
    # Find all nodes belonging to this Region (F, C, T, P, O)
    node_indices = [idx for idx, n in enumerate(node_labels) if n[0].upper() == cat]
    
    total_vals_in_region = len(node_indices) * heatmap_data.shape[0]
    
    for hemi in hemi_groups:
        hemi_indices = [idx for idx in node_indices if get_hemi(node_labels[idx]) == hemi]
        
        if hemi_indices and total_vals_in_region > 0:
            vals = np.asarray(heatmap_data)[:, hemi_indices].flatten()
            pct_30 = (np.sum(vals > 0.30) / total_vals_in_region) * 100.0
            pct_50 = (np.sum(vals > 0.50) / total_vals_in_region) * 100.0
            pct_75 = (np.sum(vals > 0.75) / total_vals_in_region) * 100.0
            pct_90 = (np.sum(vals > 0.90) / total_vals_in_region) * 100.0
        else:
            pct_30, pct_50, pct_75, pct_90 = 0.0, 0.0, 0.0, 0.0
            
        stats_data.append({'Brain Region': cat, 'Threshold': '> 0.30', 'Hemi': hemi, 'Percentage': pct_30})
        stats_data.append({'Brain Region': cat, 'Threshold': '> 0.50', 'Hemi': hemi, 'Percentage': pct_50})
        stats_data.append({'Brain Region': cat, 'Threshold': '> 0.75', 'Hemi': hemi, 'Percentage': pct_75})
        stats_data.append({'Brain Region': cat, 'Threshold': '> 0.90', 'Hemi': hemi, 'Percentage': pct_90})

df_stats = pd.DataFrame(stats_data)

# --- 4. Plot the Stacked Bar Charts ---
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 12))

colors = sns.color_palette("Set2", 3)
color_map = dict(zip(hemi_groups, colors))

x = np.arange(len(region_groups))
width = 0.35

def plot_grouped_stacked(ax, thresh1, thresh2, title):
    # thresh1
    df1 = df_stats[df_stats['Threshold'] == thresh1].pivot(index='Brain Region', columns='Hemi', values='Percentage')
    df1 = df1.reindex(index=region_groups, columns=hemi_groups).fillna(0)
    
    # thresh2
    df2 = df_stats[df_stats['Threshold'] == thresh2].pivot(index='Brain Region', columns='Hemi', values='Percentage')
    df2 = df2.reindex(index=region_groups, columns=hemi_groups).fillna(0)
    
    bottom1 = np.zeros(len(region_groups))
    bottom2 = np.zeros(len(region_groups))
    
    for hemi in hemi_groups:
        vals1 = df1[hemi].values
        vals2 = df2[hemi].values
        
        # Plot thresh1 on left
        bars1 = ax.bar(x - width/2, vals1, width, bottom=bottom1, label=hemi, color=color_map[hemi], edgecolor='white', linewidth=0.5)
        # Plot thresh2 on right (no label to avoid legend duplicates)
        bars2 = ax.bar(x + width/2, vals2, width, bottom=bottom2, color=color_map[hemi], edgecolor='white', linewidth=0.5)
        
        # Add labels
        ax.bar_label(bars1, labels=[f'{v:.1f}%' if v > 2 else '' for v in vals1], label_type='center', fontsize=8)
        ax.bar_label(bars2, labels=[f'{v:.1f}%' if v > 2 else '' for v in vals2], label_type='center', fontsize=8)
        
        bottom1 += vals1
        bottom2 += vals2

    ax.set_xticks(x)
    ax.set_xticklabels(region_groups)
    ax.set_title(title, fontsize=14)
    ax.set_ylabel('Percentage of Values (%)', fontsize=12)
    ax.set_ylim(0, 100)
    ax.legend(title='Hemisphere', loc='upper right')

# --- Plot 1: > 0.75 and > 0.90 ---
plot_grouped_stacked(ax1, '> 0.75', '> 0.90', 'Percentage of Jaccard Index Values > 0.75 (Left) and > 0.90 (Right)')

# --- Plot 2: > 0.50 and > 0.30 ---
plot_grouped_stacked(ax2, '> 0.50', '> 0.30', 'Percentage of Jaccard Index Values > 0.50 (Left) and > 0.30 (Right)')

ax2.set_xlabel('Brain Regions', fontsize=12)

plt.tight_layout()
plt.show()
```
